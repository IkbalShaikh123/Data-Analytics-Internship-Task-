```python
# ==========================================
# FUNCTIONS FOR DATA CLEANING
# ==========================================

def parse_rating(value):
    """
    Convert rating text such as
    '4.6 out of 5 stars' into 4.6
    """
    if pd.isna(value):
        return np.nan
    
    match = re.search(r'(\d+(?:\.\d+)?)', str(value))
    
    if match:
        return float(match.group(1))
    
    return np.nan


def parse_number(value):
    """
    Convert values such as:
    2,457 -> 2457
    6K+   -> 6000
    1.5M  -> 1500000
    """
    if pd.isna(value):
        return np.nan
    
    value = str(value).strip().replace(",", "")
    
    match = re.search(r'(\d+(?:\.\d+)?)([KMB]?)', value, re.IGNORECASE)
    
    if not match:
        return np.nan
    
    number = float(match.group(1))
    suffix = match.group(2).upper()
    
    multiplier = {
        "K": 1_000,
        "M": 1_000_000,
        "B": 1_000_000_000
    }
    
    return number * multiplier.get(suffix, 1)


def parse_price(value):
    """
    Convert price text into numeric value.
    Example:
    '$15.99' -> 15.99
    """
    if pd.isna(value):
        return np.nan
    
    match = re.search(r'(\d+(?:,\d{3})*(?:\.\d+)?)', str(value))
    
    if match:
        return float(match.group(1).replace(",", ""))
    
    return np.nan


print("Cleaning functions created successfully!")
```

```python
# Create a copy for cleaning

cleaned_df = df.copy()

print("Clean copy created.")
print("Original shape:", df.shape)
print("Cleaned dataframe shape:", cleaned_df.shape)
```

```python
# Rename columns

cleaned_df = cleaned_df.rename(columns={
    "Title": "product_title",
    "Rating": "product_rating",
    "Number_of_reviews": "total_reviews",
    "bought_in_last_month": "purchased_last_month",
    "Current/discounted_price": "discounted_price",
    "Listed_price": "original_price",
    "is_best_seller": "is_best_seller",
    "is_sponsored": "is_sponsored",
    "is_couponed": "has_coupon",
    "buy_box_availability": "buy_box_availability",
    "delivery_details": "delivery_details",
    "sustainability_badges": "sustainability_tags",
    "image_url": "product_image_url",
    "collected_at": "data_collected_at"
})

cleaned_df.columns
```

```python
# Convert product rating to numeric

cleaned_df["product_rating"] = cleaned_df["product_rating"].apply(parse_rating)

print(cleaned_df["product_rating"].dtype)

cleaned_df[["product_rating"]].head()
```

```python
# Convert total reviews to numeric

cleaned_df["total_reviews"] = cleaned_df["total_reviews"].apply(parse_number)

cleaned_df[["total_reviews"]].head()
```

```python
# Convert monthly purchases to numeric

cleaned_df["purchased_last_month"] = (
    cleaned_df["purchased_last_month"]
    .apply(parse_number)
)

cleaned_df[["purchased_last_month"]].head()
```

```python
# Convert discounted and original prices to numeric

cleaned_df["discounted_price"] = (
    cleaned_df["discounted_price"]
    .apply(parse_price)
)

cleaned_df["original_price"] = (
    cleaned_df["original_price"]
    .apply(parse_price)
)

cleaned_df[
    ["discounted_price", "original_price"]
].head()
```

```python
# Calculate discount percentage

cleaned_df["discount_percentage"] = np.where(
    (cleaned_df["original_price"] > 0) &
    cleaned_df["discounted_price"].notna(),
    
    ((cleaned_df["original_price"] -
      cleaned_df["discounted_price"])
     / cleaned_df["original_price"]) * 100,
    
    np.nan
)

cleaned_df["discount_percentage"] = (
    cleaned_df["discount_percentage"]
    .clip(lower=0, upper=100)
    .round(2)
)

cleaned_df[
    ["discounted_price",
     "original_price",
     "discount_percentage"]
].head(10)
```

```python
# Remove duplicate rows

before_duplicates = len(cleaned_df)

cleaned_df = cleaned_df.drop_duplicates()

after_duplicates = len(cleaned_df)

removed_duplicates = before_duplicates - after_duplicates

print("Rows before removing duplicates:", before_duplicates)
print("Rows after removing duplicates:", after_duplicates)
print("Duplicate rows removed:", removed_duplicates)
```

```python
# Check missing values before handling them

missing_before = cleaned_df.isnull().sum()

missing_before[missing_before > 0]
```

```python
# Numeric columns

numeric_columns = [
    "product_rating",
    "total_reviews",
    "purchased_last_month",
    "discounted_price",
    "original_price",
    "discount_percentage"
]

for column in numeric_columns:
    
    median_value = cleaned_df[column].median()
    
    cleaned_df[column] = (
        cleaned_df[column]
        .fillna(median_value)
    )

print("Numeric missing values handled.")
```

```python
# Categorical columns

categorical_columns = [
    "is_best_seller",
    "is_sponsored",
    "has_coupon",
    "buy_box_availability",
    "sustainability_tags"
]

for column in categorical_columns:
    
    cleaned_df[column] = (
        cleaned_df[column]
        .fillna("Unknown")
    )

print("Categorical missing values handled.")
```

```python
# Handle missing text and URL values

text_columns = [
    "product_title",
    "product_image_url",
    "delivery_details"
]

for column in text_columns:
    
    cleaned_df[column] = (
        cleaned_df[column]
        .fillna("Not Available")
    )

print("Text missing values handled.")
```

```python
# Final missing-value check

remaining_missing = cleaned_df.isnull().sum()

print("Remaining missing values:")
print(remaining_missing[remaining_missing > 0])
```
